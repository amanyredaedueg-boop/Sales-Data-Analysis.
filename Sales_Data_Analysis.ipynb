```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
total_sales = df['Sales'].sum()
print("Total Sales:", total_sales)
```

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.replace(';', '', regex=False)
```

```python
print(df.columns.tolist())
```

```python
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
```

```python
print(df['Profit'].dtype)
```

```python
total_profit = df['Profit'].sum()
print("Total Profit:", total_profit)
```

```python
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
total_sales = df['Sales'].sum()
print("Total Sales:", total_sales)
```

```python
# Calculate total sales for each product category
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
print(category_sales)
```

```python
import matplotlib.pyplot as plt
```

```python
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
subcategory_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
print(subcategory_sales)
```

```python
subcategory_sales.plot(kind='bar', figsize=(10,5))
plt.title('Sales by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_products)
```

```python
top_products.plot(kind='barh', figsize=(10,6))
plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.gca().invert_yaxis()
plt.show()
```

```python
# Extract the year from Order Date
df['Year'] = df['Order Date'].dt.year
```

```python
df['Month'] = df['Order Date'].dt.month
```

```python
# Calculate total sales for each year
yearly_sales = df.groupby('Year')['Sales'].sum()
print(yearly_sales)
```

```python
yearly_sales.plot(kind='bar')
plt.title('Sales by Year')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
print(monthly_sales)
```

```python
monthly_sales.plot(kind='line', marker='o', figsize=(10,5))
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(range(1, 13))
plt.grid()
plt.show()
```

```python
# Calculate total sales for each region
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(region_sales)
```

```python
region_sales.plot(kind='bar')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
discount_profit = df.groupby('Discount')['Profit'].sum().sort_values()
print(discount_profit)
```

```python
discount_profit.plot(kind='bar', figsize=(10,5))
plt.title('Profit by Discount')
plt.xlabel('Discount')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
best_category = category_sales.idxmax()
best_category_sales = category_sales.max()
print("Best Category:", best_category)
print("Sales:", best_category_sales)
```

```python
best_subcategory = subcategory_sales.idxmax()
best_subcategory_sales = subcategory_sales.max()
print("Best Sub-Category:", best_subcategory)
print("Sales:", best_subcategory_sales)
```

```python
best_region = region_sales.idxmax()
best_region_sales = region_sales.max()
print("Best Region:", best_region)
print("Sales:", best_region_sales)
```

```python
best_year = yearly_sales.idxmax()
best_year_sales = yearly_sales.max()
print("Best Year:", best_year)
print("Sales:", best_year_sales)
```

```python
best_product = top_products.idxmax()
best_product_sales = top_products.max()
print("Best Product:", best_product)
print("Sales:", best_product_sales)
```

```python
print("===== Sales Analysis Summary =====")
print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())
print("Best Category:", best_category)
print("Best Sub-Category:", best_subcategory)
print("Best Region:", best_region)
print("Best Year:", best_year)
print("Best Product:", best_product)
```
